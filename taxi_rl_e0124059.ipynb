```python
import random
import gym
env = gym.make('Taxi-v3', render_mode='ansi')
```

```python
env.reset()
env.render()
```

```python
q = {}
for s in range(env.observation_space.n):
    for a in range(env.action_space.n):
        q[(s,a)] = 0.0
```

```python
def update_q_table(prev_state, action, reward, nextstate, alpha, gamma):

    qa = max([q[(nextstate, a)] for a in range(env.action_space.n)])
    q[(prev_state,action)] += alpha * (reward + gamma * qa - q[(prev_state,action)])
```

```python
def epsilon_greedy_policy(state, epsilon):
    if random.uniform(0,1) < epsilon:
        return env.action_space.sample()
    else:
        return max(list(range(env.action_space.n)), key = lambda x: q[(state,x)])
```

```python
alpha = 0.4
gamma = 0.999
epsilon = 0.017
```

```python
for i in range(8000):
    r = 0
    prev_state = env.reset()
    if isinstance(prev_state, tuple):
        prev_state = prev_state[0]

    while True:
        # Select the action by epsilon-greedy policy
        action = epsilon_greedy_policy(prev_state, epsilon)

        # Perform the action and receive feedback
        step_result = env.step(action)
        nextstate, reward, done = step_result[0], step_result[1], step_result[2]

        # Update the Q value
        update_q_table(prev_state, action, reward, nextstate, alpha, gamma)

        # Update state and accumulate reward
        prev_state = nextstate
        r += reward

        if done:
            break

    if (i + 1) % 1000 == 0:
        print(f"Episode: {i + 1}, total reward: {r}")

env.close()
```
